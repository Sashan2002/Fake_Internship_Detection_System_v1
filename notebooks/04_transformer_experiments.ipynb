# 04 - Transformer Experiments (E5)

Fine-tunes BERT and RoBERTa transformer-only classifiers and compares them under identical conditions (Section 8.1, Section 10, dev sequence step 9). Requires `torch` and `transformers` (`pip install -r requirements.txt`) and, for realistic runtimes, a GPU.

```python
from ml.models.transformer_model import build_transformer_model, TransformerModelConfig
from ml.nlp.bert_model import BertTrainingConfig

config = TransformerModelConfig(architecture='bert')
training_config = BertTrainingConfig(epochs=3, batch_size=16, learning_rate=2e-5, random_seed=42)
model = build_transformer_model(config, training_config)
print(model.config)
```

## Repeat with RoBERTa for the controlled comparison

```python
from ml.nlp.roberta_model import RobertaTrainingConfig

roberta_config = TransformerModelConfig(architecture='roberta')
roberta_training_config = RobertaTrainingConfig(epochs=3, batch_size=16, learning_rate=2e-5, random_seed=42)
roberta_model = build_transformer_model(roberta_config, roberta_training_config)
print(roberta_model.config)
```
