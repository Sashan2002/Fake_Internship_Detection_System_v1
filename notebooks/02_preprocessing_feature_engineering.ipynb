# 02 - Preprocessing & Feature Engineering

Builds the 24 engineered columns (Appendix A, fields 19-42) and the duplicate-aware train/validation/test split (Section 18).

```python
import pandas as pd
from ml.preprocessing.feature_engineering import engineer_features, ENGINEERED_FEATURE_COLUMNS
from ml.preprocessing.dataset_split import compute_duplicate_group_id, duplicate_aware_split, check_group_split_integrity

df = pd.read_csv('../ml/data/raw/emscad_research_ready.csv')
df = engineer_features(df)
df[ENGINEERED_FEATURE_COLUMNS].describe()
```

## Duplicate-aware split

Per Section 18: duplicate/near-duplicate groups must not cross splits. If `duplicate_group_id` is not already present, derive it conservatively (see `compute_duplicate_group_id`'s docstring for the exact-match caveat).

```python
if 'duplicate_group_id' not in df.columns:
    df['duplicate_group_id'] = compute_duplicate_group_id(df)

split_df = duplicate_aware_split(df, random_seed=42)
offending = check_group_split_integrity(split_df)
assert offending == [], f'Leakage across splits: {offending}'
split_df['split'].value_counts()
```

## Save the processed, split dataset

```python
split_df.to_csv('../ml/data/processed/emscad_processed_split.csv', index=False)
print('Saved.')
```
